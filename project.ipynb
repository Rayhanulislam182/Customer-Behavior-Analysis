1. Missing Value Handling

```python
df['Category']
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

2. Column name rename

```python
df.columns
```

```python
df.columns=df.columns.str.lower()
```

```python
df.columns
```

```python
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':"purchase_amount"})
```

```python
df.columns
```

3. Age group create

```python
df["age_group"]=pd.qcut(df['age'],q=4,labels=['young','middle','imediate_senior','senior'])
```
